## 2. 从一个小问题开始

假设我们暂时用一条直线估计房屋价格：**预测价格 = 斜率 × 面积 + 基准价**。面积用平方米，价格单位是“示例万元”。这只是人为设定的合成规则；后续课程会学习如何从数据估计参数。

先运行下格，观察三个合成样本。

````python
# 合成数据：面积（平方米）与示例价格（万元）
面积列表 = [40, 60, 80]
示例价格列表 = [210, 290, 370]

for 面积, 价格 in zip(面积列表, 示例价格列表):
    print(f"面积 {面积:>2} 平方米，示例价格 {价格} 万元")
````

## 3. 用手工参数做预测

先人为选斜率 4、基准价 50。每增加 1 平方米，预测增加 4 个示例万元。逐条计算预测，并比较预测与合成样本。

````python
斜率 = 4
基准价 = 50
预测价格列表 = [斜率 * 面积 + 基准价 for 面积 in 面积列表]

print("面积 | 合成示例 | 手工预测 | 绝对误差")
for 面积, 真实值, 预测值 in zip(面积列表, 示例价格列表, 预测价格列表):
    print(f"{面积:>4} | {真实值:>8} | {预测值:>8} | {abs(真实值 - 预测值):>8}")
````

平均绝对误差是每个样本“预测与示例值相差多少”的绝对值平均。它保留价格单位，数值越小表示这组预测在这些样本上越接近。这里的“真实值”变量仅指给定的合成示例值。

````python
平均绝对误差 = sum(abs(真实值 - 预测值) for 真实值, 预测值 in zip(示例价格列表, 预测价格列表)) / len(示例价格列表)
print(f"平均绝对误差：{平均绝对误差:.1f} 示例万元")
````

## 4. 改一个参数，观察预测怎么变

先想一想：斜率从 4 改为 5，面积 80 平方米的预测会增加多少？运行实验观察每个样本预测的变化。我们这里只比较手工参数，没有训练算法。

````python
新斜率 = 5
新预测价格列表 = [新斜率 * 面积 + 基准价 for 面积 in 面积列表]

for 面积, 旧预测, 新预测 in zip(面积列表, 预测价格列表, 新预测价格列表):
    print(f"{面积} 平方米：原预测 {旧预测}，新预测 {新预测}，变化 {新预测 - 旧预测}")
````

因为基准价不变，预测变化等于斜率变化乘以面积。较大的面积带来更大的预测变化。试着把 `新斜率` 改为 3，再运行这一格；观察变化的方向。

## 5. 变量会记住之前的运行

内核会保留变量。先比较两种写法：第一格每次都重设起点，第二格保留此前的状态。分别重复运行，观察结果。

````python
计数 = 0
计数 += 1
print("当前计数：", 计数)
````

重复运行后计数仍显示 1，因为这格每次都先把它重设为 0。现在运行下面这格两次，它没有重设，状态会继续累加。然后按工具栏的“重启内核”并运行全部单元格，状态会重新建立。

````python
# 重复运行此格，累计值会继续变化
累计次数 = globals().get("累计次数", 0)
累计次数 += 1
print("本次累计次数：", 累计次数)
````

若跳着运行依赖前面变量的单元，可能遇到 `NameError`，也可能读到旧变量值。出现这种情况时，从头运行全部单元格，保证状态顺序一致。我们用一份不依赖隐藏状态的纯函数版本：输入相同，输出相同。

````python
def 增加一次(当前值):
    """返回加一后的新值，不修改外部变量。"""
    return 当前值 + 1

起始值 = 0
结果甲 = 增加一次(起始值)
结果乙 = 增加一次(起始值)
print("相同输入的两次结果：", 结果甲, 结果乙)
print("原始输入仍为：", 起始值)
````
